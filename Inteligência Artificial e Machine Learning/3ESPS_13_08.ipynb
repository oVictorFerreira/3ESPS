```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,11)
y = 2*x + 3 + np.random.normal(0,1.5,11)
plt.plot(x,y,'bo')
plt.show()
```

```python
def calcula_erro(a,b):
    y_p = a*x + b
    erro = 0
    for k in range(len(x)):
        erro += (y_p[k] - y[k])**2
    return erro

A = np.linspace(-10,10,100)
B = np.linspace(-10,10,100)

melhor_a = A[0]
melhor_b = B[0]
menor_erro = calcula_erro(melhor_a,melhor_b)
matriz_erros = np.zeros((len(A),len(B)))
for i,a in enumerate(A):
    for j,b in enumerate(B):
        erro = calcula_erro(a,b)
        matriz_erros[i][j] = erro
        if erro < menor_erro:
            menor_erro = erro
            melhor_a = a
            melhor_b = b

AA, BB = np.meshgrid(A, B, indexing='ij')
fig, ax = plt.subplots(subplot_kw={'projection': '3d'}, figsize=(10, 7))
surf = ax.plot_surface(AA, BB, matriz_erros, cmap='viridis', edgecolor='none')
ax.set_xlabel('Coeficiente A')
ax.set_ylabel('Coeficiente B')
ax.set_zlabel('Erro Quadrático')
ax.set_title('Superfície de Erro (Regressão Linear)')
fig.colorbar(surf, shrink=0.5, aspect=5, label='Valor do Erro')
plt.show()


y_p = melhor_a*x + melhor_b
plt.plot(x,y,'bo')
plt.plot(x,y_p)
plt.show()
```
